<span style='font-size:small'>This graph can help us visualize what data we should gather. Every flare occurs at a particular time, and at each time there is a corresponding flux.  So we can store all the values of the flux at each point at which a flare occurs to prepare for the bootstrapping and t\-test we will conduct later on. </span>



```python
# We'll store the data for the flux at the time of the flares here
fluxAtFlare = []
fluxAtFlareRetro = [] #This list will store the data that Prof. Martin had access to, the original 163 flares, so we can compare our results with his

# Iterate over each entry on the flare table and estimate the corresponding flux value based on the data
for x in range(1, len(flare_table)): # We start at 1 to avoid treating the title line as data
    closestTimeIndex = 0
    flareAfterFlux = False
    for y in range(len(spot_time)):
        # Find index that produces a spot time nearest the flare time 
        if abs((spot_time[closestTimeIndex] - flare_time[x])) > abs((spot_time[y] - flare_time[x])):
            closestTimeIndex = y
            # Check to see if the flare happened before or after the time we identify. This will be useful for interpolation 
            if spot_time[closestTimeIndex] < flare_time[x]:
                flareAfterFlux = True
            else:
                flareAfterFlux = False
    
    # Now we work out the closest times on either side of the flare.
    if flareAfterFlux == True: # If the flare happened after our closest time, we need the next time
        closestTimeBefore = spot_time[closestTimeIndex]
        closestFluxBefore = spot_flux[closestTimeIndex]
        
        closestTimeAfter = spot_time[closestTimeIndex + 1]
        closestFluxAfter = spot_flux[closestTimeIndex + 1]
        
    else: # If the flare happened before our closest time, we need the time before it
        closestTimeBefore = spot_time[closestTimeIndex - 1]
        closestFluxBefore = spot_flux[closestTimeIndex -1]
            
        closestTimeAfter = spot_time[closestTimeIndex]
        closestFluxAfter = spot_flux[closestTimeIndex]

    # Now we calculate the slope between the two points, find where the flare lies, and estimate a corresponding flux value
    interpolationSlope = (closestFluxAfter - closestFluxBefore)/(closestTimeAfter - closestTimeBefore)
    
    estimatedFlux = (flare_time[x] - closestTimeBefore)*interpolationSlope + closestFluxBefore
        
        
    fluxAtFlare.append(estimatedFlux)
    
    # Keep the first 163 flares (x = 1 to 163; row 0 is the header). These are the flares from Martin et al
    if x < 164:
        fluxAtFlareRetro.append(estimatedFlux)

print(len(fluxAtFlare))
```

## Chapter 4 \-\-\- Bootstrapping

<span style='font-size:small'>OVERVIEW:</span>

<span style='font-size:small'>**Bootstrapping**</span><span style='font-size:small'> is a statistical technique that involves resampling a single given dataset with replacement in order to create many simulated bootstrapped samples of the same size as the original dataset. For bootstrapping to be successful it is vital that we resample </span><span style='font-size:small'>**with replacement**</span><span style='font-size:small'> in order to ensure that the simulated sample mean can differ with each resample. We can then use our produced distribution of simulated samples to perform statistical analysis such as calculating standard errors and performing hypothesis tests. Bootstrapping provides us with a more accurate representation of the entire population than our smaller initial dataset, therefore improving the accuracy and reliability of our conclusions.</span>

<span style='font-size:small'><br></span>

<span style='font-size:small'>DEFINITIONS:</span>

<span style='font-size:small'>\-\-&gt; Population: The complete group about which some information is desired </span>

<span style='font-size:small'>\-\-&gt; Sample: An unbiased subset of the population</span>

<span style='font-size:small'>\-\-&gt; Null Hypothesis: A statistical hypothesis that claims that no statistical significance exists in a given set of observations</span>

<span style='font-size:small'>\-\-&gt; P\-value: The probability that the null hypothesis produces the observed result or weirder</span>

<span style='font-size:small'><br></span>

<span style='font-size:small'>OUR BOOTSTRAPPING: </span>

<span style='font-size:small'>We can analyze how correlated flare occurrence and positive spot modulation flux are by using the bootstrapping method to derive a p\-value! </span>

<span style='font-size:small'>In this case, we define our null hypothesis to be the assumption that there is </span><span style='font-size:small'>**no positive correlation between the flare occurrence and positive spot modulation flux,**</span><span style='font-size:small'> meaning that a flare is equally likely to occur at any given time, independent of the flux. Then, we use bootstrapping, where we randomly select 244 flares from all the fluxes that we have in our dataset, with replacement, to create a simulated sample of flare occurrences. By repeating this process multiple times, 10000 times to be exact, we create a distribution of simulated samples. For each simulated sample we then checked if we had the same number, or a greater number, of flares at positive fluxes as our observed sample, which was </span><span style='font-size:small'>**148/244**</span><span style='font-size:small'>. </span>

<span style='font-size:small'>We then take the number of simulated samples for which this occurs and divide it by the total number of simulated samples to obtain our </span><span style='font-size:small'>**p\-value**</span><span style='font-size:small'>! Remember, a small p\-value indicates that the probability of this outcome occurring given that the null hypothesis is true is very small. </span>

<span style='font-size:small'>**Note:**</span><span style='font-size:small'> This is a one\-tailed test because in this case we only care about the samples that have 148 or more flares at positive fluxes.</span>



```python
N = 10000
n_flares = len(fluxAtFlare)        # 244 flares in total
observed_positive = positiveFluxCount  # 148 of them have positive flux (counted earlier)

num = 0
# Collect 10,000 samples of randomly selected flares
for ii in range(N):
    # Randomly draw 244 flares (with replacement!) from all fluxes in our sample
    bootstrap = np.random.choice(spot_flux, n_flares, replace = True)
    # Count how many of the randomly selected flares have a positive flux
    flares_pos_flux = np.sum(bootstrap > 0)

    # If this many or more flares correspond to a positive flux, then we count it since it is as weird as (or weirder than) what we observed
    if flares_pos_flux >= observed_positive:
        num += 1

# The chance that we get equal to or more than the observed number of flares at positive fluxes in our simulated samples is the p-value
print('Our calculated p-value is ' + str(num/N))

# Cross-check: the bootstrap above is a simulation of a binomial test, so we can also calculate the exact answer.
# Under the null hypothesis, each flare has the same chance of being positive as a random point of the spot modulation data.
from scipy.stats import binom
chance_positive = np.mean(spot_flux > 0)
print('Fraction of all spot modulation data that is positive: ' + str(chance_positive))
print('Exact binomial p-value (should be close to the bootstrap value): ' + str(binom.sf(observed_positive - 1, n_flares, chance_positive)))
```

<span style='font-size:medium'>**Bootstrapping Conclusions:**</span>

<span style='font-size:small'>Let's sum up our findings from bootstrapping:</span>

<span style='font-size:small'>Null Hypothesis: There is </span><span style='font-size:small'>**no positive correlation**</span><span style='font-size:small'> between flare occurrence and positive spot modulation flux.</span>

<span style='font-size:small'>P\-value: ~0.009 \(the exact value changes a little each run because the resampling is random\) \-\-&gt; This p\-value tells us that there is a less than 1% chance of our observed result occurring, given that the null hypothesis is true. This is incredibly small and so we can conclude that there </span><span style='font-size:small'>_does_</span><span style='font-size:small'> exist a positive correlation between flare occurrence and positive spot modulation flux.</span>

<span style='font-size:small'>In Section 7.3.1. of Martin et al's study, they derived a p\-value of 0.0083 from 10,000 samples, which is the same amount we used. Their p\-value is similar to ours, which shows that the correlation remains present even when taking the new data into account. </span>



## Chapter 5 \-\-\- T\-test

<span style='font-size:small'>OVERVIEW:</span>

<span style='font-size:small'>An alternate method for testing this correlation is by using a </span><span style='font-size:small'>**t\-test**</span><span style='font-size:small'>. A t\-test is a statistical test that compares the means of two groups. There exists both one\-sample t\-tests and two\-sample t\-tests, and which one is implemented depends on the nature of the data and the comparison being made. </span><span style='font-size:small'>**One\-sample t\-tests**</span><span style='font-size:small'> are used when comparing the mean of a single sample to a known population mean or theoretical expectation. This helps determine whether the sample differs significantly from the expected value. </span><span style='font-size:small'>**Two\-sample t\-tests**</span><span style='font-size:small'>, on the other hand, are used when comparing the means of two independent groups to assess whether they are significantly different from each other. </span>

<span style='font-size:small'><br></span>

<span style='font-size:small'>EQUATION: </span>$t =   \frac{\overline{x} -  \mu }{ \frac{ s }{ \sqrt{n} } }$<span style='font-size:small'> </span>

<span style='font-size:small'>\-\-&gt; </span>$\overline{x}$<span style='font-size:small'> = sample mean</span>

<span style='font-size:small'>\-\-&gt; </span>$\mu$<span style='font-size:small'> = assumed/ hypothesised mean</span>

<span style='font-size:small'>\-\-&gt; </span>$s$<span style='font-size:small'> = standard deviation of sample</span>

<span style='font-size:small'>\-\-&gt; n = sample size</span>

<span style='font-size:small'><br></span>

<span style='font-size:small'>OUR T\-TEST: </span>

<span style='font-size:small'>We can analyze how correlated flare occurrence and positive spot flux are by using a t\-test to derive a p\-value! To do this, we conducted a one\-sample t\-test, comparing the distribution of the spot modulation flux of the 244 flares to a normal distribution with a mean of zero. </span>

<span style='font-size:small'>The first step is to calculate the </span><span style='font-size:small'>**t\-value**</span><span style='font-size:small'> using the equation above. The t\-value is the ratio of the difference between the mean of the sample and the hypothesised mean, and the variation that exists within the sample sets.</span>

<span style='font-size:small'>Then, as we did for our bootstrapping, we define our null hypothesis to be the assumption that there is </span><span style='font-size:small'>**no positive correlation between the flare occurrence and positive spot modulation flux,**</span><span style='font-size:small'> meaning that a flare is equally likely to occur at any given time, independent of the flux. Next, we use our calculated t\-value to find our </span><span style='font-size:small'>**p\-value**</span><span style='font-size:small'>. In a t\-test this p\-value represents the probability of observing a t\-value as, or more, extreme than the t\-value calculated from the sample data, under the assumption that the null hypothesis is true. As was the case in our bootstrap test, a small p\-value indicates that the probability of this calculated t\-value occurring given that the null hypothesis is true is very small. </span>



```python
# Student T-test 
# Compare distribution of spot fluxes distribution with mean of zero 

# A function to run a one-sample student t-test on a distribution. Assumes that the null hypothesis mean will be 0 
def studentTTest(distribution):
    n = len(distribution)
    s = np.std(distribution,ddof=1)
    standardError = s/np.sqrt(n)
    t = np.mean(distribution)/standardError
    return t


tValueRetro = studentTTest(fluxAtFlareRetro)
tValueNew = studentTTest(fluxAtFlare)

print("t value of old data:", tValueRetro)
print("t value of new data:", tValueNew)

print("Mean ± std of current data:", np.mean(fluxAtFlare), "±", np.std(fluxAtFlare, ddof=1))
print("Mean ± std of old data:", np.mean(fluxAtFlareRetro), "±", np.std(fluxAtFlareRetro, ddof=1))
```

```python
# Get some other information to compare the flare lists 
print("2023 min:", min(fluxAtFlareRetro), "2023 max:", max(fluxAtFlareRetro), "2023 mean:", np.mean(fluxAtFlareRetro), "2023 median:", np.median(fluxAtFlareRetro)) 

print("Updated min:", min(fluxAtFlare), "updated max:", max(fluxAtFlare), "updated mean:", np.mean(fluxAtFlare), "updated median:", np.median(fluxAtFlare))
```

#### <span style='font-size:medium'>**Confidence in T\-value:**</span>

<span style='font-size:small'>Now that we have obtained a t\-value from our t\-test, let's see how confident we are in it. We can tell how confident we are by calculating the p\-value of our t\-value. </span>

<span style='font-size:small'>\-\-&gt; To find a p value, we must first plot something called a </span><span style='font-size:small'>**T\-Distribution**</span>

<span style='font-size:small'>\-\-&gt; The T\-Distribution shows the spread of possible t\-values we would calculate if we were to take new samples </span>

<span style='font-size:small'>\-\-&gt; the distribution is modeled by the following function: </span>$f(x) = \frac{\Gamma\left(\frac{\nu+1}{2}\right)}{\sqrt{\nu\pi}\,\Gamma\left(\frac{\nu}{2}\right)}\left(1+\frac{x^2}{\nu}\right)^{-\frac{\nu+1}{2}}$<span style='font-size:small'> </span>

<span style='font-size:small'>\-\-&gt; </span>$\nu$<span style='font-size:small'> = degrees of freedom \(for a one\-sample t\-test, $\nu = n - 1$\)</span>

<span style='font-size:small'>\-\-&gt; </span>$\Gamma(T)$<span style='font-size:small'> = the gamma function</span>



```python
from scipy.stats import t as student_t

# A one-sample t-test on n values has n - 1 degrees of freedom
dfNew = len(fluxAtFlare) - 1
dfRetro = len(fluxAtFlareRetro) - 1

# Plot the t-distribution between -5 and 5 (further than that the values are negligible)
linSpace = np.linspace(-5,5,1000)
tDistribution = student_t.pdf(linSpace, dfNew)

# The p-value is the chance of getting a t-value at least as large as ours if the null hypothesis (mean = 0) is true.
# It is one-sided, like the bootstrap test, because we only care about whether the mean flux is ABOVE zero.
# student_t.sf(t, df) = 1 - cdf(t), calculated exactly (no need to look it up on a grid of points).
pValueRetro = student_t.sf(tValueRetro, dfRetro)
pValueNew = student_t.sf(tValueNew, dfNew)

print("2023 data (first 163 flares) p-value:", pValueRetro)
print("New data p-value:", pValueNew)

# Visual representation of the t-distribution and the t value for the new data
plt.plot(linSpace,tDistribution)
tail = linSpace >= tValueNew
plt.fill_between(linSpace[tail], tDistribution[tail], color='r', alpha=0.3, label='p-value (area to the right of T)')
plt.title('T Distribution', size=22)
plt.xlabel('T value', size=22)
plt.ylabel('PDF', size=22)
plt.vlines(tValueNew,0,0.4, colors='r', linestyles='dashed')
plt.annotate(f"Our T-value (T={tValueNew:.3f})", (tValueNew,0.4))
plt.legend()
plt.show()
```

<span style='font-size:medium'>**T\-Test Conclusions:**</span>

<span style='font-size:small'>Let's sum up our findings from the t\-test:</span>

<span style='font-size:small'>Null Hypothesis: There is </span><span style='font-size:small'>**no correlation**</span><span style='font-size:small'> between flare occurrence and positive spot modulation flux.</span>

<span style='font-size:small'>P\-value: 0.206 \-\-&gt; This p\-value tells us that there is about a 20.6% chance of seeing a t\-value at least this large \(one\-sided, since we only care whether the mean is above zero\), if we assume that the null hypothesis is true. This is not sufficient to reject the null hypothesis, and thus is not sufficient to conclude that there exists a correlation between the mean time of flare occurrence and spot modulation flux.</span>
